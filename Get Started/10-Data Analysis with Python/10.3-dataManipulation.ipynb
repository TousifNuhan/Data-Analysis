```python
## Data Aggregating and Grouping

grouped_mean=df.groupby('CONTACTLASTNAME')['SALES'].mean()
grouped_mean.head()
```

```python
grouped_mean2=df.groupby(['CONTACTFIRSTNAME','CONTACTLASTNAME'])['SALES'].mean()
print(grouped_mean2)
```

```python
grouped_sum=df.groupby(['MONTH_ID','CONTACTFIRSTNAME'])['SALES'].sum()
print(grouped_sum.head(12))
```

```python
grouped_mean3=df.groupby(['MONTH_ID','CONTACTFIRSTNAME'])['SALES'].mean()
print(grouped_mean3)
```

```python
## Aggregate multiple functions

grouped_agg=df.groupby('CONTACTFIRSTNAME')['SALES'].agg(['sum','mean','count'])  ## count: The number of orders/rows in the dataset for that contact’s first name.
print(grouped_agg)
```

```python
## Merging and Joining DataFrames

# Create simple DataFrames
df1=pd.DataFrame({'key':['A','B','C'],'Value1':[1,2,3]})
df2=pd.DataFrame({'key':['A','B','D'],'Value2':[3,4,5]})

```

```python
df1
```

```python
df2
```

```python
## Merge DataFrames on the 'Key Columns'

pd.merge(df1,df2,on='key',how='inner')  ## inner means it take only common names key
```

```python
pd.merge(df1,df2,on='key',how='outer') ## it will take all the keys but which value is not present at that place it is NAN
```

```python
pd.merge(df1,df2,on='key',how='left')
```

```python
pd.merge(df1,df2,on='key',how='right')
```
