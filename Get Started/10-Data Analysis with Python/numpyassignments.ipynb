```python
### Assignment 3: Array Operations

## 1. Create two NumPy arrays of shape (3, 4) filled with random integers. Perform element-wise addition, subtraction, multiplication, and division.

arr5=np.arange(0,12).reshape(3,4)
arr6=np.arange(12,24).reshape(3,4)
print(arr5)
print(arr6)

# addition
add=arr5+arr6
print(add)

#subtraction
sub=arr5-arr6
print(sub)

#multiplication
mult=arr5*arr6
print(mult)

#division
div=arr5/arr6
print(div)

## 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

arr7=np.arange(0,16).reshape(4,4)
print(arr7)
row_wise=arr7.sum(axis=1)   ## sums across columns for each row.
column_wise=arr7.sum(axis=0)      ## sums across rows for each columns.

print(row_wise)
print(column_wise)
```

```python
### Assignment 9: Structured Arrays

## 1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

data_type=[('name','U10'),('age','i4'),('weight','f4')]
data =np.array(
    [
        ('Nuhan', 24, 65.5),
        ('Tousif', 25, 65.8),
        ('Gazi', 22, 67.5)
    ],
    dtype=data_type
)

sorted=np.sort(data,order='age')
print(sorted)

## 2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

data_type2=[('x','i4'),('y','i4')]
points=np.array([
     (4,5),
     (7,8),
     (1,2)
],
dtype=data_type2
)

def euclidean_distance(p1, p2):
    return np.sqrt((p1['x'] - p2['x'])**2 + (p1['y'] - p2['y'])**2)

n = len(points)
distance_matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        distance_matrix[i, j] = euclidean_distance(points[i], points[j])

print("Points:\n", points)
print("\nEuclidean Distance Matrix:\n", distance_matrix)
```

```python
### Assignment 10: Masked Arrays

## 1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

import numpy as np
import numpy.ma as ma

arr=np.random.randint(1,21,size=(4,4))

mask_greater_than_10=ma.masked_greater(arr,10)
print(mask_greater_than_10)

sum_unmasked=mask_greater_than_10.sum()
print(sum_unmasked)

## 2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.
np.random.seed(1)
arr2=np.random.randint(1,21,size=(3,3))
print(arr2)
masked_diagonal=ma.masked_array(arr2,mask=np.eye(3,dtype=bool))

print(masked_diagonal)

unMasked_mean=masked_diagonal.mean()
print(unMasked_mean)

filled_array=masked_diagonal.filled(unMasked_mean)
print(filled_array)

```
