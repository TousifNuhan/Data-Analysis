```python
import pandas as pd
from io import StringIO

# The outer [ ... ] → makes it an array (so Pandas sees it as tabular).
Data = """
[
{
  "employee_id": 101,
  "first_name": "Alice",
  "last_name": "Johnson",
  "email": "alice.johnson@example.com",
  "phone": "+1-202-555-0147",
  "job_title": "Software Engineer",
  "department": "IT",
  "hire_date": "2021-06-15",
  "salary": 75000,
  "address": {
    "street": "123 Maple Street",
    "city": "San Francisco",
    "state": "CA",
    "zip": "94105",
    "country": "USA"
  },
  "skills": ["Python", "SQL", "Machine Learning"],
  "manager_id": 56
}]"""

df=pd.read_json(StringIO(Data))
```
