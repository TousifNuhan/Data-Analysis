```python
### Assignment 3: Data Cleaning

## 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

df=pd.DataFrame(np.random.randint(1,50,size=(5,3)),columns=["A","B","C"])
print(df)

## NAN values
df.iloc[0,1]=np.nan
df.iloc[1,2]=np.nan
df.iloc[2,1]=np.nan

print("\nwith NAN:")
print(df)

df.fillna(df.mean(),inplace=True)
print(df)

# 2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

print("\nDF2: ")
df2=pd.DataFrame(np.random.randint(1,100,size=(6,4)))
print(df2)

df2.iloc[0,0]=np.nan
df2.iloc[1,1]=np.nan
df2.iloc[2,2]=np.nan
# print("\niloc:\n")
print(df2)

df2.dropna(inplace=True)
print("\nDrop rows with NAN: ")
print(df2)

```

```python
### Assignment 4: Data Aggregation

## 1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

df=pd.DataFrame({'Category':np.random.choice(['A','B','C'],size=8),'Value':np.random.randint(1,11,size=8)})
print(df)

group_dataframe= df.groupby('Category')['Value'].agg(['sum','mean'])
print(group_dataframe)

## 2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

df=pd.DataFrame({'Product':np.random.choice(['Product1','Product2','Product3'],size=9),'Category':np.random.choice(['A','B','C'],size=9),'Sales':np.random.randint(1,11,9)})
print("DataFrame2: \n",df)

group_category2=df.groupby('Category')['Sales'].sum()
print(group_category2)
```

```python
### Assignment 5: Merging DataFrames

## 1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

df1=pd.DataFrame(
    {
        'Key':['A','B','C','D'],
        'Value1':np.random.randint(1,21,size=4)
    }
)

print("DF2: \n",df1)

df2=pd.DataFrame(
    {
        'Key':['A','B','C','G'],
        'Value2':np.random.randint(1,21,size=4)
    }
)
print("DF2: \n",df2)

merged=pd.merge(df1,df2,on='Key',how='inner')

print(merged)

## 2. Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.

df1=pd.DataFrame({
    'A':np.random.randint(1,11,size=4),
    'B':np.random.randint(11,21,size=4)
}) 

print("DF1: \n",df1)

df2=pd.DataFrame({
    'C':np.random.randint(21,31,size=4),
    'D':np.random.randint(31,41,size=4)
})
print("DF2: \n",df2)

concat_row=pd.concat([df1,df2],axis=0)
print("Concat row: \n",concat_row)

concat_col=pd.concat([df1,df2],axis=1)
print("Concat Column: \n",concat_col)
```
