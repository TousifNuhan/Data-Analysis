```python
import requests

base_url="https://pokeapi.co/api/v2/"

def get_pokemon_info(name):
    url=f"{base_url}/pokemon/{name}"
    response=requests.get(url)
    print(response)

    if response.status_code == 200:
        poke_data = response.json()
        return poke_data
    else:
        print(f"error",response.status_code)
     

poke_name="ditto"
poke_info=get_pokemon_info(poke_name)
print(poke_info)

poke_info['name']
```

```python
import requests
import pandas as pd
from io import BytesIO

# API URL
url = "https://www.amarstock.com/data/download/CSV"

# Example payload (update if API supports date range)
payload = {
    "param1": "value1",
    "param2": "value2"
}

# Cookies and headers
cookies = {
    ".AspNet.ApplicationCookie": "YOUR_COOKIE_HERE"
}
headers = {"User-Agent": "Mozilla/5.0", "Content-Type": "application/x-www-form-urlencoded"}

# Make the POST request
response = requests.post(url, data=payload, headers=headers, cookies=cookies)

if response.status_code == 200:
    # Read CSV into pandas DataFrame
    df = pd.read_csv(BytesIO(response.content))

    # Filter for Janata Bank
    janata_df = df[df['Scrip'].str.contains('JANATA', case=False)]

    # Filter for date range Feb 2024 to Aug 30, 2025
    janata_df['Date'] = pd.to_datetime(janata_df['Date'], format='%Y%m%d')
    start_date = '2024-02-01'
    end_date = '2025-08-30'
    janata_df = janata_df[(janata_df['Date'] >= start_date) & (janata_df['Date'] <= end_date)]

    # Save filtered data
    janata_df.to_csv("janata_bank_data.csv", index=False)
    print("Filtered CSV saved for Janata Bank!")
else:
    print("Failed to download CSV. Status code:", response.status_code)
    print(response.text)
```

```python
import requests
import pandas as pd
import time
from datetime import datetime, date, time, timedelta

startDate = '2025-01-01'
formatDate = datetime.strptime(startDate, '%Y-%m-%d')
unadjData = pd.DataFrame(columns = ['Date', 'Ticker', 'Open', 'High', 'Low', 'Close', 'Volume'])

while True:
  if str(formatDate.date()) == '2025-09-01':
    print('Collection Complete')
    break
  print(str(formatDate.date()))
  url = 'https://www.amarstock.com/data/download/CSV'

  myobj = {'QuotesType': '0',
           'date': str(formatDate.date())}
  x = requests.post(url, data = myobj,timeout=30)

  if x.status_code != 200:
    print('Failed to collect data for', str(formatDate.date()))
  else:
    # print(x.text)
    data = x.text.split('\n')

    for row in data:
      if row == 'Date,Scrip,Open,High,Low,Close,Volume':
        continue
      rowVal = row.split(',')
      rowVal[0] = rowVal[0][0] + rowVal[0][1] + rowVal[0][2] + rowVal[0][3] + '-' + rowVal[0][4] + rowVal[0][5] + '-' + rowVal[0][6] + rowVal[0][7]
      unadjData.loc[len(unadjData)] = rowVal
  formatDate += timedelta(days = 1)
unadjData.to_csv('amarstock_2024_2024.csv', index = False)


# from google.colab import files
# files.download('UnAdjusted-AmarStock.csv')
```
