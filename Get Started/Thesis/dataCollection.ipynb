```python
import pandas as pd
import requests

def amarstock_download_full(symbol, start=None, end=None, to_csv=False, csv_filename=None):
    """
    Download full historical data from AmarStock for a stock, with optional CSV export.

    Parameters:
    -----------
    symbol : str
        Stock trading code, e.g. 'ZAHEENSPIN'
    start : str (YYYY-MM-DD), optional
        Start date for filtering
    end : str (YYYY-MM-DD), optional
        End date for filtering
    to_csv : bool, default False
        Whether to save the DataFrame to a CSV file
    csv_filename : str, optional
        Filename for the CSV export (if None, defaults to '{symbol}_EOD.csv')

    Returns:
    --------
    DataFrame : Full historical stock data
    """
    base_url = f"https://www.amarstock.com/stock/{symbol}"
    all_data = []
    page = 1

    while True:
        url = f"{base_url}?page={page}"
        response = requests.get(url)
        if response.status_code != 200:
            print(f"Page {page} not found, stopping.")
            break
        
        # Extract tables from the page
        tables = pd.read_html(response.text)
        
        if len(tables) < 3:
            print(f"No historical table found on page {page}.")
            break
        
        df_page = tables[2]
        if df_page.empty:
            print(f"No data on page {page}, stopping.")
            break
        
        # Clean column names
        df_page.columns = [c.strip() for c in df_page.columns]
        df_page['Date'] = pd.to_datetime(df_page['Date'], errors='coerce')
        df_page = df_page.dropna(subset=['Date'])
        
        all_data.append(df_page)
        page += 1

    if not all_data:
        print("No historical data found for this stock.")
        return None

    # Concatenate all pages
    df = pd.concat(all_data)
    df.set_index('Date', inplace=True)
    df.sort_index(inplace=True)

    # Filter by start/end dates
    if start:
        df = df[df.index >= pd.to_datetime(start)]
    if end:
        df = df[df.index <= pd.to_datetime(end)]

    # Export to CSV if requested
    if to_csv:
        if not csv_filename:
            csv_filename = f"{symbol}_EOD.csv"
        df.to_csv(csv_filename, index=True)
        print(f"Data saved to {csv_filename}")

    return df

# Example usage:
df_full = amarstock_download_full(
    "ZAHEENSPIN",
    start="2024-02-21",
    end="2025-08-21",
    to_csv=True
)

print(df_full.head())
print(df_full.tail())
print(df_full.shape)
```
